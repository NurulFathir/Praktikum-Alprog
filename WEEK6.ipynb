```python
#A. DATAFRAME SEBAGAI ARRAY NUMPY UMUM
#Objek Dataframe Pandas
#Dataframe sbg array Numpy umum
import pandas as pd
import pandas as pd
luas_dict = {
    'Indonesia': 1905,
    'Malaysia': 330,
    'Singapura': 0.7,
    'Thailand': 513,
    'Vietnam': 331
}

luas = pd.Series(luas_dict)

penduduk_dict = {
    'Indonesia': 270,
    'Malaysia': 32,
    'Singapura': 5.7,
    'Thailand': 69,
    'Vietnam': 97
}
penduduk = pd.Series(penduduk_dict)
penduduk
```

```python
Negara = pd.DataFrame({
    'Luas': luas,
    'Penduduk': penduduk
})

Negara
```

```python
Negara.index
```

```python
Negara.columns
```

```python
#B. DataFrame sebagai Dictionary Khusus
#Dataframe sebagai kamus khusus

Negara['Luas']
```

```python
#C. Membangun objek DataFrame
#1. Dari objek series tunggal
pd.DataFrame(luas, columns=['Luas'])
```

```python
#2. Dari sebuah List dictionary
pd.DataFrame({'luas': luas, 'penduduk': penduduk})
```

```python
#2 Daro array NumPy dua dimendi
pd.DataFrame(np.random.rand(5, 5),
             columns=['A', 'B', 'C', 'D', 'E'],
             index=['Baris 1', 'Baris 2', 'Baris 3', 'Baris 4', 'Baris 5'])
```
