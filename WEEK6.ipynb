OBJEK DATAFRAME PANDAS

```python
#A. DATAFRAME SEBAGAI ARRAY NUMPY UMUM
# Objek Dataframe Pandas
# Dataframe sbg Array Numpy Umum
import numpy as np
import pandas as pd

luas_dict = {'Brasil': 8515,
             'Argentina': 2780,
             'Meksiko': 1964,
             'Peru': 1285,
             'Kolombia': 1141,
             'Bolivia': 1098,
            }

luas = pd.Series(luas_dict)

penduduk_dict = {'Brasil': 2158,
                 'Argentina': 4512,
                 'Meksiko': 1278,
                 'Peru': 3245,
                 'Kolombia': 5100,
                 'Bolivia': 1180,
                }

penduduk = pd.Series(penduduk_dict)
penduduk
```

```python
Negara  = pd.DataFrame({'Luas': luas, 
                       'Penduduk': penduduk})
Negara
```

```python
Negara.index
```

```python
Negara.columns
```

```python
#B. DataFrame Sebagai Dictionary Khusus
#DataFrame sebagai Kamus Khusus

Negara['Luas']
```

```python
#C. Membangun Objek DataFrame
#1. Dari Objek Series Tunggal

pd.DataFrame(luas, columns=['Luas'])
```

```python
#2. Dari Sebuah List Dictionary
pd.DataFrame({'luas': luas, 'penduduk': penduduk})
```

```python
#3. Dari array NumPy dua dimensi
pd.DataFrame(np.random.rand(4, 4),
             columns=['p', 'q', 'r', 's'],
             index=['W', 'X', 'Y', 'Z'])
```

```python
#B. Indeks sebagai Set yang Dipesan
# Indeks sebagai set yang dipesan
obj1 = pd.Index([3, 6, 9, 12])
obj2 = pd.Index([3, 5, 7, 9])

obj1 & obj2 #intersection
```

```python
#intersect
irisan = obj1.intersection(obj2)
irisan
```

```python
obj1 & obj2 #intersection
gabungan = obj1.union(obj2)
gabungan
```

```python
obj1 | obj2 #union
```

```python
obj1 ^ obj2 #symmetric difference
```

Seleksi Data Dalam DataFrame

```python
# Membuat Series Luas
Luas = pd.Series({'Brasil': 8515,
                  'Argentina': 2780,
                  'Meksiko': 1964,
                  'Peru': 1285,
                  'Kolombia': 1141,
                  'Bolivia': 1098,
                 })

# Membuat Series Penduduk
penduduk = pd.Series({'Brasil': 2158,
                      'Argentina': 4512,
                      'Meksiko': 1278,
                      'Peru': 3245,
                      'Kolombia': 5100,
                      'Bolivia': 1180,
                     })

# Menggabungkan keduanya ke dalam DataFrame
data = pd.DataFrame({'Luas': Luas, 'Penduduk': penduduk})
data
```
